# Tutorial sobre Funciones de Activación de Redes Neuronales

¡Bienvenido a este tutorial sobre funciones de activación de redes neuronales! Las funciones de activación son componentes cruciales en las redes neuronales que introducen no linealidad, permitiendo a la red aprender patrones complejos.

En esta notebook, exploraremos algunas de las funciones de activación más comunes, entenderemos sus definiciones matemáticas, visualizaremos su comportamiento y veremos ejemplos numéricos sencillos.

## 1. Función Sigmoide (Función Logística)

La función Sigmoide es una de las funciones de activación más antiguas y ampliamente utilizadas. Mapea cualquier número real a un valor entre 0 y 1, lo que la hace adecuada para problemas de clasificación binaria donde la salida necesita ser interpretada como una probabilidad.

**Fórmula:**
$$ \sigma(x) = \frac{1}{1 + e^{-x}} $$

**Derivada:**
La derivada de la función Sigmoide se utiliza a menudo en la retropropagación:
$$ \sigma'(x) = \sigma(x) * (1 - \sigma(x)) $$

### Implementación y Gráfico

```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    s = sigmoid(x)
    return s * (1 - s)

# Generar valores de x
x = np.linspace(-5, 5, 100)

# Calcular sigmoide y su derivada
y_sigmoid = sigmoid(x)
y_sigmoid_prime = sigmoid_derivative(x)

# Graficar
plt.figure(figsize=(10, 6))
plt.plot(x, y_sigmoid, label='Función Sigmoide')
plt.plot(x, y_sigmoid_prime, label='Derivada Sigmoide', linestyle='--')
plt.title('Función de Activación Sigmoide y su Derivada')
plt.xlabel('x')
plt.ylabel('Salida')
plt.axvline(0, color='gray', linestyle='--', linewidth=0.8)
plt.axhline(0, color='gray', linestyle='--', linewidth=0.8)
plt.axhline(1, color='gray', linestyle='--', linewidth=0.8)
plt.legend()
plt.grid(True)
plt.show()

# ### Ejemplo Numérico

# Veamos cómo la función Sigmoide transforma un único valor de entrada.

input_value = 2.5
sig_output = sigmoid(input_value)
sig_deriv_output = sigmoid_derivative(input_value)

print(f"\nEjemplo Numérico para Sigmoide:")
print(f"Entrada (x): {input_value:.2f}")
print(f"Sigmoide(x): {sig_output:.4f}")
print(f"Derivada Sigmoide(x): {sig_deriv_output:.4f}")

# La función Sigmoide comprime la entrada en el rango (0, 1). Su derivada es más alta alrededor de x=0, lo que indica dónde la función es más sensible a los cambios.

# ---
```

## 2. Función ReLU (Unidad Lineal Rectificada)

ReLU es una de las funciones de activación más populares en el aprendizaje profundo. Es computacionalmente eficiente y ayuda a mitigar el problema del gradiente desvanecido, lo que la convierte en una buena opción para las capas ocultas en muchas redes neuronales.

**Fórmula:**
$$ ReLU(x) = max(0, x) $$

**Derivada:**
$$ ReLU'(x) = \begin{cases} 1 & \text{si } x > 0 \\ 0 & \text{si } x < 0 \\ \text{indefinida} & \text{si } x = 0 \end{cases} $$
(Para propósitos prácticos en la retropropagación, la derivada en x=0 a menudo se toma como 0 o 1.)

### Implementación y Gráfico

```python
def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return np.where(x > 0, 1, 0)

# Generar valores de x
x = np.linspace(-5, 5, 100)

# Calcular ReLU y su derivada
y_relu = relu(x)
y_relu_prime = relu_derivative(x)

# Graficar
plt.figure(figsize=(10, 6))
plt.plot(x, y_relu, label='Función ReLU')
plt.plot(x, y_relu_prime, label='Derivada ReLU', linestyle='--')
plt.title('Función de Activación ReLU y su Derivada')
plt.xlabel('x')
plt.ylabel('Salida')
plt.axvline(0, color='gray', linestyle='--', linewidth=0.8)
plt.axhline(0, color='gray', linestyle='--', linewidth=0.8)
plt.legend()
plt.grid(True)
plt.show()

# ### Ejemplo Numérico

# Veamos cómo la función ReLU transforma un único valor de entrada.

input_value_positive = 2.5
input_value_negative = -1.0
relu_output_pos = relu(input_value_positive)
relu_deriv_output_pos = relu_derivative(input_value_positive)
relu_output_neg = relu(input_value_negative)
relu_deriv_output_neg = relu_derivative(input_value_negative)

print(f"\nEjemplo Numérico para ReLU:")
print(f"Entrada (x): {input_value_positive:.2f}")
print(f"ReLU(x): {relu_output_pos:.4f}")
print(f"Derivada ReLU(x): {relu_deriv_output_pos:.4f}")

print(f"\nEntrada (x): {input_value_negative:.2f}")
print(f"ReLU(x): {relu_output_neg:.4f}")
print(f"Derivada ReLU(x): {relu_deriv_output_neg:.4f}")

# ReLU emite la entrada directamente si es positiva, y cero en caso contrario. Su derivada es 1 para entradas positivas y 0 para entradas negativas, lo que simplifica los cálculos de gradiente.

# ---
```

## 3. Función Tanh (Tangente Hiperbólica)

Tanh es otra función de activación popular que comprime los valores de entrada a un rango entre -1 y 1. Está centrada en cero, lo que puede facilitar el entrenamiento de redes neuronales que con la función Sigmoide, ya que ayuda a equilibrar los gradientes.

**Fórmula:**
$$ tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}} $$

**Derivada:**
$$ tanh'(x) = 1 - (tanh(x))^2 $$

### Implementación y Gráfico

```python
def tanh(x):
    return np.tanh(x)

def tanh_derivative(x):
    t = tanh(x)
    return 1 - t**2

# Generar valores de x
x = np.linspace(-5, 5, 100)

# Calcular Tanh y su derivada
y_tanh = tanh(x)
y_tanh_prime = tanh_derivative(x)

# Graficar
plt.figure(figsize=(10, 6))
plt.plot(x, y_tanh, label='Función Tanh')
plt.plot(x, y_tanh_prime, label='Derivada Tanh', linestyle='--')
plt.title('Función de Activación Tanh y su Derivada')
plt.xlabel('x')
plt.ylabel('Salida')
plt.axvline(0, color='gray', linestyle='--', linewidth=0.8)
plt.axhline(0, color='gray', linestyle='--', linewidth=0.8)
plt.axhline(-1, color='gray', linestyle='--', linewidth=0.8)
plt.axhline(1, color='gray', linestyle='--', linewidth=0.8)
plt.legend()
plt.grid(True)
plt.show()

# ### Ejemplo Numérico

# Veamos cómo la función Tanh transforma un único valor de entrada.

input_value = 1.8
tanh_output = tanh(input_value)
tanh_deriv_output = tanh_derivative(input_value)

print(f"\nEjemplo Numérico para Tanh:")
print(f"Entrada (x): {input_value:.2f}")
print(f"Tanh(x): {tanh_output:.4f}")
print(f"Derivada Tanh(x): {tanh_deriv_output:.4f}")

# La función Tanh comprime la entrada en el rango (-1, 1). Al igual que Sigmoide, su derivada es más alta alrededor de x=0, pero está centrada en cero, lo que puede ser ventajoso.

# ---
```

## 4. Función Softmax

La función Softmax se utiliza a menudo en la capa de salida de una red neuronal para problemas de clasificación multiclase. Toma un vector de puntuaciones de valores reales arbitrarios y los transforma en una distribución de probabilidad, donde cada valor está entre 0 y 1, y todos los valores suman 1.

**Fórmula:**
Para un vector $z = [z_1, z_2, ..., z_K]$ de K números reales, la función Softmax se define como:
$$ \text{Softmax}(z_i) = \frac{e^{z_i}}{\sum_{j=1}^{K} e^{z_j}} $$

**Derivada:**
La derivada de la función Softmax es más compleja ya que depende de todos los elementos del vector de entrada. Para $i=j$:
$$ \frac{\partial \text{Softmax}(z_i)}{\partial z_j} = \text{Softmax}(z_i) (1 - \text{Softmax}(z_j)) $$
Y para $i \neq j$:
$$ \frac{\partial \text{Softmax}(z_i)}{\partial z_j} = -\text{Softmax}(z_i) \text{Softmax}(z_j) $$

Para visualizar la función Softmax y sus derivadas de forma análoga a las funciones escalares anteriores, consideraremos un caso simplificado donde solo una de las entradas de un vector varía, mientras las otras se mantienen constantes. Graficaremos las probabilidades resultantes y sus derivadas con respecto a esta entrada variable.

### Implementación, Gráfico y Ejemplo Numérico

```python
def softmax(x):
    # Restar np.max(x) para estabilidad numérica y evitar desbordamientos
    e_x = np.exp(x - np.max(x, axis=-1, keepdims=True))
    return e_x / np.sum(e_x, axis=-1, keepdims=True)

# Generar valores de x (para una entrada variable)
x_values = np.linspace(-5, 5, 100)

# Crear un vector de entrada donde solo la primera componente varía, las otras son constantes (e.g., 0)
# Para un ejemplo de 3 clases: [x, 0, 0]
input_vectors = np.array([[val, 0, 0] for val in x_values])

# Calcular la salida de Softmax para cada vector de entrada
softmax_outputs = softmax(input_vectors)

# Extraer las probabilidades para cada clase
p0 = softmax_outputs[:, 0]
p1 = softmax_outputs[:, 1]
p2 = softmax_outputs[:, 2]

# Calcular las derivadas de las probabilidades con respecto a la primera entrada (x)
# Para dp_i/dx donde x es la primera entrada (z_0):
# dp0/dx = p0 * (1 - p0)
# dp1/dx = -p1 * p0
# dp2/dx = -p2 * p0
dp0_dx = p0 * (1 - p0)
dp1_dx = -p1 * p0
dp2_dx = -p2 * p0

# Graficar la función Softmax
plt.figure(figsize=(12, 6))
plt.plot(x_values, p0, label='P(Clase 0) = Softmax(x)', color='blue')
plt.plot(x_values, p1, label='P(Clase 1) = Softmax(0)', color='red')
plt.plot(x_values, p2, label='P(Clase 2) = Softmax(0)', color='green')
plt.title('Función Softmax (cuando la primera entrada varía)')
plt.xlabel('Valor de la primera entrada (x)')
plt.ylabel('Probabilidad de la clase')
plt.axvline(0, color='gray', linestyle='--', linewidth=0.8)
plt.axhline(0, color='gray', linestyle='--', linewidth=0.8)
plt.axhline(1, color='gray', linestyle='--', linewidth=0.8)
plt.legend()
plt.grid(True)
plt.show()

# Graficar las derivadas de Softmax
plt.figure(figsize=(12, 6))
plt.plot(x_values, dp0_dx, label='d(P(Clase 0))/dx', color='blue', linestyle='--')
plt.plot(x_values, dp1_dx, label='d(P(Clase 1))/dx', color='red', linestyle='--')
plt.plot(x_values, dp2_dx, label='d(P(Clase 2))/dx', color='green', linestyle='--')
plt.title('Derivadas de la Función Softmax con respecto a la primera entrada (x)')
plt.xlabel('Valor de la primera entrada (x)')
plt.ylabel('Derivada')
plt.axvline(0, color='gray', linestyle='--', linewidth=0.8)
plt.axhline(0, color='gray', linestyle='--', linewidth=0.8)
plt.legend()
plt.grid(True)
plt.show()

# ### Ejemplo Numérico

# Veamos cómo la función Softmax transforma un vector de valores de entrada.

input_vector = np.array([3.0, 1.0, 0.2])
softmax_output = softmax(input_vector)

print(f"\nEjemplo Numérico para Softmax:")
print(f"Vector de entrada (z): {input_vector}")
print(f"Salida Softmax: {softmax_output}")
print(f"Suma de las salidas Softmax: {np.sum(softmax_output):.4f}")

# Como puedes ver, la función Softmax convierte las puntuaciones de entrada brutas en una distribución de probabilidad. El valor de entrada más grande obtiene la probabilidad más alta.

# ---
```

## 5. Tabla Comparativa de Funciones de Activación

A continuación, se presenta una tabla que resume las características clave de las funciones de activación que hemos cubierto:

| Característica         | Sigmoide (Sigmoid)                        | ReLU (Rectified Linear Unit)                 | Tanh (Tangente Hiperbólica)                 | Softmax                                          |
| :--------------------- | :---------------------------------------- | :------------------------------------------- | :------------------------------------------ | :----------------------------------------------- |
| **Rango de Salida**    | (0, 1)                                    | [0, $\infty$)                                | (-1, 1)                                     | [0, 1] (suma total a 1)                         |
| **Centrado en Cero**   | No                                        | No                                           | Sí                                          | No (pero la salida es una distribución de probabilidad) |
| **No Lineal**          | Sí                                        | Sí                                           | Sí                                          | Sí                                               |
| **Diferenciable**      | Sí (en todo el dominio)                   | Sí (excepto en x=0)                          | Sí (en todo el dominio)                     | Sí (Jacobiana)                                   |
| **Problema de Vanishing Gradient** | Sí (para valores extremos de x)           | Sí (para x < 0, "neuronas muertas")        | Sí (para valores extremos de x)             | Menos propenso en la capa de salida, pero aún posible con gradientes muy pequeños |
| **Problema de Exploding Gradient** | Menos propenso                            | Propenso (para valores grandes de x)         | Menos propenso                              | Menos propenso                                   |
| **Costo Computacional**| Alto (debido a exponenciales)             | Bajo (simple comparación)                   | Alto (similar a Sigmoide)                   | Alto (debido a exponenciales y sumatorias)      |
| **Uso Común**          | Capa de salida para clasificación binaria | Capas ocultas (el más popular actualmente) | Capas ocultas (alternativa a Sigmoide)      | Capa de salida para clasificación multiclase    |

Esta tabla sirve como una referencia rápida para elegir la función de activación adecuada según los requisitos de tu red neuronal y el problema que estés resolviendo.
